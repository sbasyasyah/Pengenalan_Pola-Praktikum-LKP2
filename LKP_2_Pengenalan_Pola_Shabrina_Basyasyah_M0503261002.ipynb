```python
# Read picture
babon = cv2.imread("jinu.png", cv2.IMREAD_COLOR)

# Konversi BGR -> RGB untuk menampilkan warna asli
babon_rgb = cv2.cvtColor(babon, cv2.COLOR_BGR2RGB)

# Konversi grayscale
babon_gray = cv2.cvtColor(babon, cv2.COLOR_BGR2GRAY)

# Resize
babon_resized = cv2.resize(babon_gray, (100, 100))

# Blur
babon_blurred = cv2.GaussianBlur(babon_gray, (9, 9), 0)

# Canny Edge Detection
babon_edges = cv2.Canny(babon_gray, 100, 200)


# Menampilkan hasil
plt.figure(figsize=(15, 6))

# Original

plt.subplot(1, 6, 1)

plt.title("Original")

plt.imshow(babon_rgb)

plt.axis("off")


# Viridis

plt.subplot(1, 6, 2)

plt.title("Viridis")

plt.imshow(babon_gray, cmap="viridis")

plt.axis("off")
# Grayscale
plt.subplot(1, 6, 3)
plt.title("Grayscale")
plt.imshow(babon_gray, cmap="gray")
plt.axis("off")

# Resize
plt.subplot(1, 6, 4)
plt.title("Resized 100x100")
plt.imshow(babon_resized, cmap="gray")
plt.axis("off")

# Blur
plt.subplot(1, 6, 5)
plt.title("Blur")
plt.imshow(babon_blurred, cmap="gray")
plt.axis("off")

# Canny
plt.subplot(1, 6, 6)
plt.title("Canny Edges")
plt.imshow(babon_edges, cmap="gray")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")

# Teks awal
kalimat = (
    "EXO merupakan boyband asal Korea Selatan yang beranggotakan Xiumin, Suho, Chanyeol, Baekhyun, Chen, Kyungsoo, Kai, Sehun, dan Lay."
)

# Mengubah menjadi huruf kecil
kalimat_kecil = kalimat.lower()

# Tokenisasi
token_kata = word_tokenize(kalimat_kecil)

# Menghilangkan tanda baca
kata_bersih = [kata for kata in token_kata if kata.isalnum()]

# Menghitung jumlah kata
total_kata = len(kata_bersih)

# Output
print("Teks Asli        :", kalimat)
print("Huruf Kecil      :", kalimat_kecil)
print("Hasil Tokenisasi :", token_kata)
print("Tanpa Tanda Baca :", kata_bersih)
print("Jumlah Kata      :", total_kata)
```
